# Introduction

Experiment with PyTorch and in particular with Tensors.

They are similar to NumPy's ndarrays, but Tensors can run on GPUs or other hardware accelerators.

Tensors are also optimised for Automatic Differentiation.

```python
# Import Standard Libraries
import torch
import numpy as np
```

# Definition

## Integer Tensors

```python
# 1-D Tensor or Vector
tensor = torch.tensor([1, 2, 3])
```

```python
tensor
```

```python
# 2-D Tensor or Matrix
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])
```

```python
matrix
```

## Float Tensors

```python
# Create a float tensor
tensor_float = torch.FloatTensor([1, 2, 3])
tensor_float_2 = torch.tensor([1, 2, 3], dtype=torch.float)
```

```python
tensor_float
```

```python
tensor_float_2
```

## Double Tensors

```python
# Create a double tensor
tensor_double = torch.DoubleTensor([1, 2, 3])
tensor_double_2 = torch.tensor([1, 2, 3], dtype=torch.double)
```

```python
tensor_double
```

```python
tensor_double_2
```

```python
# Convert integer to double
tensor_integer = torch.tensor([1, 2, 3], dtype=torch.int)
print(tensor_integer.to(torch.double))
```

## Empty

```python
torch.empty(3, 4)
```

```python
torch.zeros(3, 4)
```

```python
torch.ones(3, 4)
```

## Tensor Like

```python
base = torch.tensor([1, 2, 3])
```

```python
print(torch.empty_like(base))
print(torch.zeros_like(base))
print(torch.ones_like(base))
```

# Shape

```python
# Shape and size() give the same results
print(tensor.shape)
print(matrix.shape)
print(tensor.size())
print(matrix.size())
```

# Statistics

```python
# Get the mean and std
tensor_double.mean()
```

```python
tensor_double.std()
```

# Reshape

```python
# It is done through the method 'view'
tensor_double.view(-1, 1)
```

```python
# The '-1' means that the dimension has to be inferred from the second one
matrix.view(-1, 2)
```

# Copying

```python
base = torch.tensor([1, 2, 3, 4])
copy = base.clone()
print(copy)
```

# Random

```python
# Random numbers from normal distribution
torch.randn(4, 4)
```

```python
# Random integer numbers between 1 and 10
torch.randint(1, 10, (4, 2))
```

```python
# Generate a random tensor with the same size of another one
tensor_random = torch.randn(4, 4)
print(torch.randn_like(tensor_random, dtype=torch.double))
```

# Operations

```python
# Add
torch.add(torch.tensor([1, 2, 3], dtype=torch.double), 
          torch.tensor([4, 5, 6], dtype=torch.double))
```

# GPU Usage

## General Accelerator

### Move to Device

```python
if torch.accelerator.is_available():
    print('Accelerator is available: ', torch.accelerator.current_accelerator())
    tensor = tensor.to(torch.accelerator.current_accelerator())
```

### Create Already on the Device

```python
my_device = torch.accelerator.current_accelerator() if torch.accelerator.is_available() else torch.device('cpu')
print('Device: {}'.format(my_device))
```

```python
x = torch.rand(2, 2, device=my_device)
print(x)
```
